#Load and chunk ALL PDFs

```python
import os

doc_path = '/content/drive/MyDrive/Healthcare_AI_RAG_Assistant/FDA Drug Levels PDF'
all_chunks = []

for filename in os.listdir(doc_path):
    if filename.endswith('.pdf'):
        loader = PyPDFLoader(os.path.join(doc_path, filename))
        pages = loader.load()
        doc_chunks = text_splitter.split_documents(pages)
        all_chunks.extend(doc_chunks)
        print(f"{filename}: {len(doc_chunks)} chunks")

print(f"\nTotal chunks from all documents: {len(all_chunks)}")
```

#Generate embeddings and store them in Chroma

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore = Chroma(
    persist_directory="/content/drive/MyDrive/Healthcare_AI_RAG_Assistant/chroma_db_local",
    embedding_function=embeddings
)

existing_count = vectorstore._collection.count()
print(f"Existing vectors: {existing_count}")

if existing_count == 0:
    vectorstore = Chroma.from_documents(
        documents=all_chunks,
        embedding=embeddings,
        persist_directory="/content/drive/MyDrive/Healthcare_AI_RAG_Assistant/chroma_db_local"
    )
    print("Created new vector store")
else:
    print("Loaded existing vector store — no re-embedding needed")
```

#Build retrieval (test search before adding the LLM)

```python
query = "What are the side effects of ibuprofen?"
results = vectorstore.similarity_search(query, k=3)

for i, doc in enumerate(results):
    print(f"--- Result {i+1} ---")
    print(f"Source: {doc.metadata.get('source', 'unknown')}")
    print(doc.page_content[:300])
    print()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=api_key
)

def ask_question(query, k=3):
    results = vectorstore.similarity_search(query, k=k)

    context = "\n\n".join([
        f"[Source: {doc.metadata.get('source', 'unknown').split('/')[-1]}]\n{doc.page_content}"
        for doc in results
    ])

    prompt = f"""Answer the question using ONLY the context below. If the context doesn't contain the answer, say so. Cite which source document(s) you used.

Context:
{context}

Question: {query}

Answer:"""

    response = llm.invoke(prompt)
    return response.content if isinstance(response.content, str) else response.content[0]['text']

answer = ask_question("What are the side effects of ibuprofen?")
print(answer)
```

#adding SQLite for document metadata

```python
import sqlite3
import os
from datetime import datetime

db_path = '/content/drive/MyDrive/Healthcare_AI_RAG_Assistant/project_metadata.db'
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

cursor.execute('''
CREATE TABLE IF NOT EXISTS documents (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    filename TEXT NOT NULL,
    filepath TEXT NOT NULL,
    chunk_count INTEGER,
    file_size_kb REAL,
    date_processed TEXT
)
''')

conn.commit()
print("Table 'documents' created successfully")
```

populate it with actual document data

```python
doc_path = '/content/drive/MyDrive/Healthcare_AI_RAG_Assistant/FDA Drug Levels PDF'

cursor.execute('DELETE FROM documents')

for filename in os.listdir(doc_path):
    if filename.endswith('.pdf'):
        filepath = os.path.join(doc_path, filename)

        loader = PyPDFLoader(filepath)
        pages = loader.load()
        doc_chunks = text_splitter.split_documents(pages)
        chunk_count = len(doc_chunks)

        file_size_kb = os.path.getsize(filepath) / 1024

        cursor.execute('''
        INSERT INTO documents (filename, filepath, chunk_count, file_size_kb, date_processed)
        VALUES (?, ?, ?, ?, ?)
        ''', (filename, filepath, chunk_count, round(file_size_kb, 2), datetime.now().isoformat()))

conn.commit()
print("Metadata inserted for all documents")
```

Verify it worked by a real SQL query

```python
cursor.execute('SELECT filename, chunk_count, file_size_kb FROM documents ORDER BY chunk_count DESC')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
from datetime import datetime
import time

for test in test_questions:
    try:
        answer = ask_question(test["question"])
    except Exception as e:
        if "RESOURCE_EXHAUSTED" in str(e) or "429" in str(e):
            print("Hit daily quota — stopping here.")
            break
        else:
            raise e

    print(f"Q: {test['question']}")
    print(f"A: {answer[:300]}")
    print("---")

    cursor.execute('''
    INSERT INTO evaluations (question, expected_behavior, actual_answer, passed, date_tested)
    VALUES (?, ?, ?, ?, ?)
    ''', (test["question"], test["expected_behavior"], answer, "TBD", datetime.now().isoformat()))

    conn.commit()
    time.sleep(2)

print("\nAll questions tested and logged")
```

```python
import time

def chat_response(message, history):
    try:
        results = vectorstore.similarity_search(message, k=3)

        context = "\n\n".join([
            f"[Source: {doc.metadata.get('source', 'unknown').split('/')[-1]}]\n{doc.page_content}"
            for doc in results
        ])

        prompt = f"""Answer the question using ONLY the context below. If the context doesn't contain the answer, say so. Cite which source document(s) you used.

Context:
{context}

Question: {message}

Answer:"""

        response = llm.invoke(prompt)
        full_answer = response.content if isinstance(response.content, str) else response.content[0]['text']

        partial = ""
        for word in full_answer.split(" "):
            partial += word + " "
            yield partial
            time.sleep(0.02)

    except Exception as e:
        if "RESOURCE_EXHAUSTED" in str(e) or "429" in str(e):
            yield "Daily API quota reached for this demo. Please try again later, or the developer can upgrade the API plan for continuous availability."
        else:
            yield f"An error occurred: {str(e)}"
```

```python
import gradio as gr

custom_css = """
.gradio-container .prose h1 {
    text-align: center;
    font-size: 2.2em !important;
}
#description-text {
    text-align: center;
    font-size: 1.8em !important;
    color: #555;
    margin-bottom: 1em;
}
.message-wrap {
    font-size: 1.5em !important;
}
textarea {
    font-size: 1.5em !important;
}
"""

description_html = """
<div id="description-text">
Ask anything about FDA drug labels — get grounded, source-cited answers, not guesses.
</div>
"""

with gr.Blocks(css=custom_css) as demo:
    gr.Markdown("# Medic AI")
    gr.HTML(description_html)
    gr.ChatInterface(
        fn=chat_response,
        examples=[
            "What is metformin used for?",
            "What are the side effects of ibuprofen?",
            "What are the warnings for lisinopril during pregnancy?"
        ],
        cache_examples=False
    )

demo.launch(share=True, debug=True)
```
